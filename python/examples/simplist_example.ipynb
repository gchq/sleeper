## Setup Sleeper Client
Setup the logger for the notebook and create a Sleeper Client that is connected to a Sleeper instance.

This assumes that Sleeper has already been deployed with a table created and partitioned. 

````python
import logging

from sleeper import SleeperClient, enable_logging

logging.basicConfig(
    level=logging.INFO,
    format="[NOTEBOOK] %(asctime)s %(levelname)s %(message)s",
    force=True,
)

jupyter_logger = logging.getLogger(__name__)

jupyter_logger.info("Starting notebook")

# Enable Sleeper Client logging level
enable_logging(logging.DEBUG)

table_name = "my-table"  # Set this here so we can use it later without running all cells.
instance_id = "instance-123"

sleeper_client = SleeperClient(instance_id)
````

## Write Data
Write a single batch of data to Sleeper

````python
rows_list = [{"key": "aaa", "value": "v1"}, {"key": "bbb", "value": "v2"}, {"key": "ccc", "value": "v3"}]

sleeper_client.write_single_batch(table_name, rows_list)
````

````python
# Look up a result using a single key
jupyter_logger.info("Performing exact key query...")

exact_results = sleeper_client.exact_key_query(table_name, {"key": ["aaa", "ccc"]})

jupyter_logger.info(f"Exact Results: {exact_results}")

# And retrieve a key range
jupyter_logger.info("Performing range key query...")

range_results = sleeper_client.range_key_query(table_name, [{"key": ["a", True, "c", False]}])

jupyter_logger.info(f"Range Results: {range_results}")
````
